```python
gerente = []
def mostrar_bienvenida():
    print("\n" + "="*70)
    print(" ✨ CEREBRO IA v2.0: MÓDULO DE DIRECCIÓN Y ESTRATEGIA DE GERENTES ✨")
    print("="*70)

def mostrar_menu():
    print("\n¿Qué acción deseas que ejecute la IA?")
    print("1. ➕ Registrar nuevo Gerente (Entrada de datos)")
    print("2. 📋 Mostrar reporte de Gerentes (Formato Tabla)")
    print("3. 🧠 Consultar a la IA (Análisis de Autoridad y Bonos)")
    print("4. ❌ Salir del sistema")
    print("-" * 70)

def registrar_gerente():
    print("\n--- REGISTRO DE NUEVA CLAVE DE GERENCIA ---")

    nombre = input("Nombre del Gerente: ")
    unidad_negocio = input("Unidad de Negocio (Ej. Innovación, Logística, Finanzas): ")
    sucursal = input("Sucursal asignada (Ej. Norte, Centro, Virtual): ")
    estrategia = input("Estrategia principal (Ej. Expansión, Reducción de costos): ")
    reuniones_diarias = input("¿Tiene reuniones diarias? (Sí/No): ")

    print("\n[ Configuración de Niveles para la IA ]")
    print("1: Alto (CEO/Director) | 2: Medio (Regional) | 3: Operativo (Local)")
    nivel_autoridad = int(input("Selecciona el nivel de autoridad (1, 2 o 3): "))

    bonos = float(input("Monto de bonos por desempeño sugerido (USD): "))
    diccionario_gerentes = {
        "Nombre": nombre,
        "Unidad": unidad_negocio,
        "Sucursal": sucursal,
        "Estrategia": estrategia,
        "Reuniones": reuniones_diarias,
        "Autoridad": nivel_autoridad,
        "Bonos": bonos
    }

    gerentes.append(diccionario_gerentes)
    print(f"\n✅ ¡IA: Datos consolidados para el Gerente '{nombre}'!")
def mostrar_tabla():
    print("\n" + "="*101)
    print(f"{'GERENTE':<12} | {'UNIDAD':<12} | {'SUCURSAL':<10} | {'AUTORIDAD':<10} | {'BONOS':<10} | {'REUNIONES':<9} | {'ESTRATEGIA PRINCIPAL':<25}")
    print("="*101)

    if len(gerentes) == 0:
        print(f"{'[ No hay registros en el sistema ]':^115}")
    else:
        for g in gerentes:
            if g['Autoridad'] == 1:
                txt_autoridad = "1-Alto"
            elif g['Autoridad'] == 2:
                txt_autoridad = "2-Medio"
            else:
                txt_autoridad = "3-Operativo"

            print(f"{g['Nombre']:<12} | {g['Unidad']:<12} | {g['Sucursal']:<10} | {txt_autoridad:<10} | ${g['Bonos']:<9.2f} | {g['Reuniones']:<9} | {g['Estrategia']:<25}")
    print("="*101)
def consultar_ia():
    print("\n--- ANALÍTICA AVANZADA DE LA IA ---")
    if len(gerentes) == 0:
        print("\u2728 IA: Base de datos vacía. No puedo procesar estrategias.")
    else:
        print("Selecciona el algoritmo de análisis:")
        print("a) Evaluar carga de trabajo (Reuniones vs Estrategia)")
        print("b) Auditoría de Bonos de Alto Nivel")
        opcion_ia = input("Selecciona (a/b): ")
        if opcion_ia == "a":
            print("\n\u2728 Dictamen de la IA sobre Gestión de Tiempo:")
            for g in gerentes:
                if g["Reuniones"].lower() == "sí" or g["Reuniones"].lower() == "si":
                    print(f"-> {g['Nombre']} en {g['Sucursal']}: Alto enfoque operativo. Su estrategia ('{g['Estrategia']}') requiere supervisión constante.")
                else:
                    print(f"-> {g['Nombre']} en {g['Sucursal']}: Enfoque autónomo. Ejecuta la estrategia ('{g['Estrategia']}') con libertad de agenda.")
        elif opcion_ia == "b":
            print("\n\u2728 Dictamen de la IA sobre Presupuesto e Incentivos:")
            for g in gerentes:
                if g["Autoridad"] == 1 and g["Bonos"] > 2000:
                    print(f"-> ⚠️ {g['Nombre']} (Nivel Alto) recibe bonos top (${g['g_bono'] if 'g_bono' in locals() else g['Bonos']}). Requiere aprobación de junta directiva.")
                elif g["Autoridad"] == 3 and g["Bonos"] > 1000:
                    print(f"-> ⚠️ {g['Nombre']} (Nivel Operativo) tiene bonos por encima del promedio local. Verificar KPIs.")
                else:
                    print(f"-> ✅ {g['Nombre']} cuenta con un esquema de bonos equilibrado para su unidad de {g['Unidad']}.")
        else:
            print("❌ Opción de análisis no reconocida por la IA.")
mostrar_bienvenida()
while True:
    mostrar_menu()
    opcion = input("Selecciona una opción del menú (1-4): ")
    if opcion == "1":
        registrar_gerente()
    elif opcion == "2":
        mostrar_tabla()
    elif opcion == "3":
        consultar_ia()
    elif opcion == "4":
        print("\n✨ IA: Sincronizando datos en la nube... Sistema cerrado correctamente. ¡Hasta pronto!")
        break
    else:
        print("\n❌ Error: Entrada inválida. Elige una opción correcta.")
print("Programa realizado por Oscar Flores NL = 22")
```
